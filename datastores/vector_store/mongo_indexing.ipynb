```python
import os
import re
from langchain_core.documents import Document

docs_path = "/Users/wnowogorski/PycharmProjects/CHAT_AGH/web_scraping/downloads"

documents = []
for filename in os.listdir(docs_path):
    url = filename.strip(".md").replace("_", "/")

    with open(os.path.join(docs_path, filename), "r", encoding="utf-8") as f:
        text = f.read()

    documents.append(
        Document(
            page_content=text,
            metadata={
                "url": url
            },
        )
    )

print(len(documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

chunks = []

t = []

splitter = RecursiveCharacterTextSplitter(chunk_size=2048)
for document in documents:
    metadata = document.metadata

    chunked_content = splitter.split_text(document.page_content)

    if chunked_content in t:
        continue
    else:
        t.append(chunked_content)

    for i, c in enumerate(chunked_content):
        chunks.append(Document(
            page_content=c,
            metadata=metadata | {"sequence_number": i}
        ))
```
