```python
from dotenv import load_dotenv
from rag.utils.utils import load_json_data

from rag.models.google_genai_models import GraphExtractorModel

ENV_PATH = "/Users/wnowogorski/PycharmProjects/CHAT_AGH/.env"
DATA_PATH = "/Users/wnowogorski/PycharmProjects/CHAT_AGH/data"
NUM_RETRIES = 3

load_dotenv(dotenv_path=ENV_PATH)
```

```python
graph_extractor = GraphExtractorModel()
```

```python
data = load_json_data(DATA_PATH)
```

```python
import time
import json

entities = []
relationships = []
srate = []

for x, file in enumerate(data):
    print(f"Processing file: {x + 1} / {len(data)}, successfully processed files: {sum(srate)}")

    success = 0
    delay = 2
    for i in range(NUM_RETRIES):
        try:
            res = graph_extractor.generate(file.page_content)
            entities.extend(res["entities"])
            relationships.extend(res["relationships"])
            success = 1
            break
        except Exception as e:
            print(e)
            time.sleep(delay)
            delay *= 5

    srate.append(success)

    with open("entities.json", "w") as f:
        json.dump(entities, f)

    with open("relationships.json", "w") as f:
        json.dump(relationships, f)

```
